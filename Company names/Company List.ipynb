```python

def save_to_csv(companies: list[str], filename: str) -> None:
    with open(filename, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["company"])
        for company in companies:
            writer.writerow([company])

save_to_csv(companies, "companies.csv")
```
