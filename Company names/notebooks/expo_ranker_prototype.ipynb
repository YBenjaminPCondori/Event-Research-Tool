# Expo Relevance Ranking Prototype

This notebook uses your existing scraped exhibitor data and ranks which stands are most relevant to visit first.

```
from pathlib import Path
import sys

# Notebook is inside /notebooks, so add parent folder to imports
PROJECT_DIR = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_DIR))

from expo_ranker import score_exhibitors, export_outputs, ScoreConfig, DEFAULT_KEYWORDS, CATEGORY_WEIGHTS
```

## 1. Define your target profile

Edit this text depending on the event or role you care about.

```
target_profile = """
machine learning, artificial intelligence, edge ai, embedded systems, fpga,
firmware, robotics, computer vision, sensors, python, c++, linux, graduate,
internship, software engineering, data, automation
""".strip()
```

## 2. Score all exhibitors

```
config = ScoreConfig(
    keywords=DEFAULT_KEYWORDS,
    category_weights=CATEGORY_WEIGHTS,
    target_profile=target_profile,
)

ranked, keyword_summary = score_exhibitors(config=config)
ranked.head(20)
```

## 3. Show the most useful columns

```
columns = [
    'rank', 'company', 'stand_or_location', 'final_score', 'relevance',
    'tfidf_similarity_0_100', 'matched_keywords'
]
ranked[columns].head(25)
```

## 4. Export ranked outputs

```
ranked_path, visit_plan_path, keyword_summary_path = export_outputs(ranked, keyword_summary)
print('Saved:')
print(ranked_path)
print(visit_plan_path)
print(keyword_summary_path)
```
